Este notebook ejecuta el orquestador con una tarea que exige dos dominios de especialización (búsqueda en la documentación y cálculo) y una síntesis final. Las salidas quedan guardadas en el notebook, así que se pueden revisar sin ejecutarlo.

## 2. La tarea

```python
print(TAREA_DEMO)
```

## 3. Ejecución

Cada decisión del supervisor, con su evaluación según la rúbrica, y cada aporte de los especialistas, con las herramientas que usaron.

```python
traza = await ejecutar(TAREA_DEMO)
```

## 4. Resumen del flujo de delegación

```python
for d in traza['decisiones']:
    print(f"Paso {d['paso']}: supervisor -> {d['siguiente']}")
print()
for c in traza['contribuciones']:
    print(f"Paso {c['paso']}: {c['agente']} usó {', '.join(c['herramientas']) or 'ninguna herramienta'}")
```
